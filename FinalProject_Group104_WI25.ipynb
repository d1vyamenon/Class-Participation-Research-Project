# Abstract

A study performed by researchers in the Czech Republic found that ninth graders who participated in class discussion tended to perform better on reading literacy tests. There is evidence that participation in class promotes learning, yet students in lecture halls tend to shy away from participation. We are curious whether academic demographic factors such as year in school, degree pursued, and field of study influence the frequency of hand raising among UCSD undergraduate students. Additionally, we are curious about what reasons cause students to abstain from raising their hand. Our data consists of the responses by UCSD undergraduate students to a survey that we created. Our data gathered demographic factors about participants as well as the degree to which certain reasons influence a student's decision to not raise their hand. Ultimately, we did not find any statistically significant differences hand-raising tendencies across students of different academic demographics. However, between frequent and infrequent hand-raisers, we noticed a difference in the most prominent reasons as to why they don't raise their hands. Infrequent hand-raisers are more heavily deterred for fear-related reasons compared to frequent hand-raisers.

## Background and Prior Work

Many university lecture halls can seat hundreds of students, but when professors ask questions during a lecture, it is not uncommon for students to sit quietly in their seats, staring at their screens or looking nervously around the room. Why is this the case, when classroom interaction is known to promote learning? A study done by researchers from the Czech Republic found that ninth graders who talked and reasoned more in class performed better on a reading literacy test.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) Although much work has been done to study student engagement, there has been very little research on student hand-raising specifically.

We wanted to explore reasons why UCSD undergraduate students decide not to raise their hands in lecture. Hand-raising is a sign of student cognitive engagement, so we decided to ask about a variety of different reasons students might not raise their hands.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) In 2003, researchers examined how classroom technology might encourage UCSD students to participate in classes.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) However, this is the only study conducted on student classroom engagement at UCSD that we were able to find.

### Null Hypotheses
There is no correlation between demographic factors and students not raising their hands in class at UCSD. Every reason is equally influential across frequent and in-frequent hand raisers at UCSD.
### Alternative Hypotheses
Newer students and Bachelor of Science (B.S.) students raise their hands less frequently at UCSD. Fear-motivated reasons are the strongest deterrents of hand-raising at UCSD.

We have inferred that newer students will raise their hands less frequently at UCSD due to being in a newer, unfamiliar environment. We believe that B.S. students at UCSD will raise their hands less frequently because the course material may be less likely to be open to interpretation. We believe that fear-motivated reasons will be the strongest deterrent of hand-raising at UCSD due to intimidating lecture size and fear of judgement.

## Analysis of Demographic Data

Our graphs in this section show significant differences across the varying demographics, but we are also aware that certain demographics are underrepresented while others are overrepresented. To determine if our data actually yields statistically significant differences we will use chi-squared tests on each of the demographic variables (year, degree, and field of study). 

```python
#Don't delete this yet!! but maybe later
df_handfreq = df.copy()
df_handfreq['hand_freq'] = df_handfreq['hand_freq'].map({'Never/Almost never': 'infrequent', 'Rarely (less than 5 times a quarter)': 'infrequent', 'Often (more than 5 times a quarter)': 'frequent', 'Every class/Almost every class ': 'frequent'})
#df_handfreq['hand_freq'].value_counts()
```

```python
year_counts = df.groupby(by=['year', 'hand_freq']).count()[['degree']].rename(columns={'degree': 'count'})
year_counts = year_counts.reset_index().pivot(index='year', columns='hand_freq', values='count')
#year_counts
chi2, p, dof, expected = stats.chi2_contingency(year_counts.T)
print("p-value for years: ", p)
```

```python
deg_counts = df_handfreq.groupby(by=['degree', 'hand_freq']).count()[['fear_talk']].rename(columns={'fear_talk': 'count'})
deg_counts = deg_counts.reset_index().pivot(index='degree', columns='hand_freq', values='count')
deg_counts = deg_counts.fillna(0)
#deg_counts
chi2, p, dof, expected = stats.chi2_contingency(deg_counts.T)
print("p-value for degrees: ", p)
```

```python
fos_counts = df_exploded.groupby(by=['field_of_study', 'hand_freq']).count()[['prof_judge']].rename(columns={'prof_judge': 'count'})
fos_counts = fos_counts.reset_index().pivot(index='field_of_study', columns='hand_freq', values='count')
fos_counts = fos_counts.fillna(0)
#fos_counts
chi2, p, dof, expected = stats.chi2_contingency(fos_counts.T)
print("p-value for field of study: ", p)
```

The p-values we obtained from each chi-squared test were above 0.05. So, although our graphs showed differences across the demographics, these differences are not statistically significant. 

### Understanding Why (Reasons)

Next, let's look at each of the different reasons why students choose not to raise their hands in lecture. To do this, we will create a new dataframe with answers to our reason questions (given on a Likert scale of "Never" to "Often/Always") from ordinal+string type data to numeric data. Then, we will take the average to represent the overall answer to each question.

```python
# Making a deep copy of the original dataframe and then converting text values to numbers
num_df = df.copy(deep = True)
response_mapping = {'Never': 0, 'Rarely': 1, 'Sometimes': 2, 'Often/Always': 3}
num_df.iloc[:,5:] = num_df.iloc[:,5:].map(lambda x: response_mapping[x])
#num_df.head(2)
```

```python
# Plotted with help from chatgpt
columns_to_convert = ['fear_talk', 'prof_judge', 'peer_judge', 'no_slow', 'no_int', 'cultural',
    'doubt', 'pace', 'unprepared', 'lost', 'internet', 'peer_ask', 'no_ans',
    'no_wrong']

# Calculate the average for each of the specified columns
average_values = num_df[columns_to_convert].mean()

# Sort the average values in ascending order
average_values_sorted = average_values.sort_values(ascending=False)

# Plot the averages using Seaborn
plt.figure(figsize=(12, 5))
ax = sns.barplot(y=average_values_sorted.index,
                 x=average_values_sorted.values,
                 hue=average_values_sorted.index,
                 palette='Spectral_r')

# Customize the plot
plt.title('Fear of Getting a Question Wrong is the Most Common Reason for Not Hand-Raising',
          fontsize=16, x=0.3, pad=20)  # Title alignment and padding
plt.ylabel('Reasons', fontsize=14)
plt.xlabel('Average Answer Value', fontsize=14)

# Adjust x-axis range
plt.xlim(0, 3)  # Set the range of the x-axis from 0 to 3 (for values between 0-3)

# Set the number of ticks and custom tick labels on the x-axis
ticks = np.linspace(0, 3, num=4)  # Create 4 ticks (0, 1, 2, 3)
tick_labels = ['Never (0)', 'Rarely (1)', 'Sometimes (2)', 'Often/Always (3)']
plt.xticks(ticks=ticks, labels=tick_labels, fontsize=12)

# Add the actual mean values on top of the bars
for i in range(len(average_values_sorted)):
    ax.text(x=average_values_sorted.iloc[i],
            y=i,
            s=f'{average_values_sorted.iloc[i]:.2f}',
            ha='left',
            va='center',
            fontsize=12,
            color='black')

# Show the plot
plt.tight_layout()
plt.show()
```

Above, you can see that the top 3 most prevalent self-reported reasons for not hand-raising are fear of getting an answer wrong (`no_wrong`, 2.09), fear of talking in front of the class (`fear_talk`, 1.92), and fear of being judged by peers (`peer_judge`, 1.86).

After running our chi-squared test on the demographic data, we ended up with very high p values that were above .05. This means our findings are statistically insignificant, and we fail to reject the first part of our null hypothesis: There is no correlation between demographic factors and students not raising their hands in class at UCSD. The relationship was likely found to be statistically insignificant simply due to the low number of survey responses. We got less than 300 responses, which simply does not represent the student body of UCSD. Furthermore, within the data, we had very skewed demographics. This will be further discussed in the limitations portion of our conclusion.

Since our subquestion was more about noticing differences than quantifying differences, we decided only to look at the values across groups and their highest-ranked reasons. This means that we can reject our null hypothesis that there is no difference in influential reasons between frequent and infrequent hand-raisers, since there is a numerical difference between groups. 

The significance values for the reasons that were highest among infrequent hand-raisers were all higher (i.e. more influential) than the values for the same reasons among frequent hand-raisers. Additionally, looking at this data, there are a lot of interesting differences in reason significance across frequency groups worth noting and possibly exploring further in future studies. For example, among frequent hand-raisers, the `no_slow` reason of not wanting to slow a lecture down had a much higher score than `fear_talk` (fear of talking in front of the class). Although `fear_talk` was ranked very high by infrequent hand-raisers, it was ranked relatively low by frequent hand-raisers, and much lower compared to the fear of slowing a lecture down.

This affects our data since it is comprised of interested students who are willing to do additional work for their fellow students, skewing our demographic data.

## Our Takeaway

Students who self-reported being frequent hand-raisers had quite different reasons for not raising their hands compared to students who self-reported being infrequent hand-raisers. For infrequent hand raisers, the primary motivation for not raising their hand are fear-based reasons. However, while fear-based reasons do contribute to why frequent hand-raisers don't raise their hands, being lost during lecture or simply not knowing the answer to a question are also large contributors. This difference in responses was quite surprising to us and allowed us to reject the second part of our null hypothesis. 
This information can also be quite helpful since many professors are always wanting more ways to increase class engagement, especially in larger lectures. This information be shared with professors and they can use it to adapt the way they interact with they students if they so choose.
